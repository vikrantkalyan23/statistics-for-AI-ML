## 1. Natural, whole, integer, and real numbers

| Set | Meaning | Examples |
|---|---|---|
| Natural $\mathbb N$ | Counting numbers; here we start at 1 | $1,2,3,\ldots$ |
| Whole | Nonnegative integers | $0,1,2,\ldots$ |
| Integer $\mathbb Z$ | Whole numbers and their negatives | $-3,0,7$ |
| Rational $\mathbb Q$ | $p/q$ for integers $p,q$, with $q\ne0$ | $-2,3/4,0.125$ |
| Real $\mathbb R$ | All points on the number line | Rational numbers and irrational numbers such as $\sqrt2,\pi$ |

Some books include 0 in $\mathbb N$; check the convention. The containment is $\mathbb N\subset\mathbb Z\subset\mathbb Q\subset\mathbb R$. An irrational number cannot be written as a ratio of integers; its decimal expansion neither terminates nor eventually repeats.

**Statistics connection:** a count is often a nonnegative integer; temperature and model weights are modeled as real numbers. Python `int` stores integers exactly, while `float` represents a finite approximation to a real number. Division by zero is undefined.

```python
from fractions import Fraction
from decimal import Decimal
import math

print("integer:", -7, "real approximation:", math.sqrt(2))
print("ordinary division:", 7 / 2, "floor division:", 7 // 2)
print("floor division rounds toward minus infinity:", -7 // 2)
```

## 2. Fractions and decimals

For $b,d\ne0$:

$$\frac ab+\frac cd=\frac{ad+bc}{bd},\qquad
\frac ab\frac cd=\frac{ac}{bd},\qquad
\frac ab\div\frac cd=\frac{ad}{bc}\quad(c\ne0).$$

Reduce a fraction by dividing numerator and denominator by their greatest common divisor: $18/24=3/4=0.75$. A reduced fraction has a terminating decimal exactly when its denominator has only factors 2 and 5; otherwise its decimal repeats.

**Worked example:** $1/3+1/6=2/6+1/6=3/6=1/2$. Multiplication by $1/2$ halves a quantity, whereas division by $1/2$ doubles it.

Most decimal fractions are not exact in binary floating point. Compare approximate results using a tolerance (`math.isclose`), rather than assuming exact equality. Construct `Fraction` and `Decimal` from integers or strings when decimal exactness matters.

```python
print("Exact sum:", Fraction(1, 3) + Fraction(1, 6))
print("Reduced:", Fraction(18, 24))
print("Float sum:", 0.1 + 0.2)
print("Close to 0.3:", math.isclose(0.1 + 0.2, 0.3))
print("Decimal sum:", Decimal("0.1") + Decimal("0.2"))
assert Fraction(1, 3) + Fraction(1, 6) == Fraction(1, 2)
```

## 3. Ratios and proportions

A ratio compares quantities: $a:b=a/b$ when $b\ne0$. Ratios should compare compatible units, or explicitly carry units (for example, kilometers per hour). Simplify $12:8$ to $3:2$.

A proportion states two ratios are equal:
$$\frac ab=\frac cd\iff ad=bc\quad(b,d\ne0).$$

**Worked example:** if 3 notebooks cost 120 rupees at a fixed unit price, 5 cost $120\times5/3=200$ rupees. This assumes direct proportionality and no fixed charge or discount.

For inverse proportionality, $xy=k$: doubling workers halves time only if total work is fixed and workers are equally productive without coordination overhead.

**Dataset split:** a train:test ratio of $4:1$ gives training fraction $4/(4+1)=0.8$, not $4/1$.

```python
n = 250
train_ratio, test_ratio = 4, 1
train_size = n * train_ratio // (train_ratio + test_ratio)
print("Train/test counts:", train_size, n - train_size)
print("Cost of 5 notebooks:", 120 * 5 / 3)
```

## 4. Percentages

A percentage is a fraction out of 100: $p\%=p/100$.

- Percentage of a quantity: $(p/100)x$.
- Percentage change from a **nonzero** old value: $(\text{new}-\text{old})/\text{old}\times100\%$. With negative baselines, interpretation needs care.
- Increase by $p\%$: multiply by $1+p/100$; decrease by $p\%$: multiply by $1-p/100$.
- Repeated changes multiply their factors; they do not generally add.

**Worked example:** accuracy rises from 80% to 88%. The absolute increase is **8 percentage points**; the relative increase is $(88-80)/80=10\%$. A 20% increase followed by a 20% decrease gives $1.2\times0.8=0.96$, a net 4% decrease.

Averaging percentages requires attention to denominators: 9/10 and 50/100 together give 59/110, not the unweighted average of 90% and 50%.

```python
old, new = 0.80, 0.88
print("Percentage-point increase:", round(100 * (new - old), 2))
print("Relative percent increase:", round(100 * (new - old) / old, 2))
print("Pooled accuracy:", (9 + 50) / (10 + 100))
print("After +20% then -20%:", 100 * 1.2 * 0.8)
```

## 5. Order of operations (BODMAS)

**Brackets → Orders (powers/roots) → Division and Multiplication → Addition and Subtraction.** Multiplication and division have equal precedence and are evaluated left to right; addition and subtraction likewise.

**Worked example:** $18/3\times2+4^2=6\times2+16=28$. Use parentheses to express your intent; avoid ambiguous inline notation such as `1/2x`.

In Python, exponentiation is `**`; `^` is bitwise XOR. Powers associate right to left: `2**3**2` means $2^{(3^2)}$. Unary minus applies after powers: `-3**2` is $-(3^2)=-9$.

```python
print(18 / 3 * 2 + 4**2)
print("Unary minus:", -3**2, "parenthesized:", (-3)**2)
print("Power association:", 2**3**2)
```

## 6. Powers and exponents

For positive real bases (and real exponents), the laws are:

$$a^ma^n=a^{m+n},\quad a^m/a^n=a^{m-n},\quad
(a^m)^n=a^{mn},\quad (ab)^n=a^nb^n.$$

For $a\ne0$, $a^0=1$ and $a^{-n}=1/a^n$. For $a>0$, $a^{1/n}=\sqrt[n]a$ and $a^{m/n}=\sqrt[n]{a^m}$. Negative bases need care with noninteger powers over the reals; for example, $(-1)^{1/2}$ is not real. The expression $0^0$ requires a context/convention; do not apply the nonzero-base rule blindly.

**Worked example:** $2^3\times2^{-5}=2^{-2}=1/4$. A squared error $(y-\hat y)^2$ is nonnegative and penalizes large errors more strongly than small errors.

## 7. Square roots

For $x\ge0$, $\sqrt x$ is the **nonnegative** number whose square is $x$. Thus $\sqrt9=3$, while solving $u^2=9$ gives $u=\pm3$.

$$\sqrt{ab}=\sqrt a\sqrt b\ (a,b\ge0),\qquad \sqrt{x^2}=|x|.$$

In general, $\sqrt{a+b}\ne\sqrt a+\sqrt b$. A root reverses a square only with the correct sign convention. Standard deviation is the square root of variance, restoring the original measurement units.

```python
print("Exponent laws:", 2**3 * 2**(-5))
print("Root of a square:", math.sqrt((-5)**2))
errors = [-2, 1, 3]
mse = sum(e**2 for e in errors) / len(errors)
print("MSE:", mse, "RMSE:", math.sqrt(mse))
assert math.isclose(mse, 14 / 3)
```

## 8. Logarithms and exponential functions

For $b>0$, $b\ne1$, and $x>0$:
$$\log_b x=y\iff b^y=x.$$

Common choices are $\log_{10}$, $\log_2$, and the natural logarithm $\ln=\log_e$, where $e\approx2.71828$. Python `math.log(x)` means $\ln x$.

For positive $x,y$:
$$\log_b(xy)=\log_b x+\log_b y,\quad
\log_b(x/y)=\log_b x-\log_b y,\quad
\log_b(x^r)=r\log_b x,\quad
\log_b x=\frac{\ln x}{\ln b}.$$

There is **no** rule $\log(x+y)=\log x+\log y$. Real logarithms of zero and negative numbers are undefined. For $b>1$, the logarithm increases; for $0<b<1$, it decreases.

The natural exponential $e^x$ is positive for all real $x$, increases, and is the inverse of $\ln x$: $\ln(e^x)=x$ and $e^{\ln x}=x$ for $x>0$. The model $A(t)=A_0e^{kt}$ grows for $k>0$ and decays for $k<0$.

**Worked example:** $\log_2 32=5$. If $A(t)=100e^{-0.2t}$, then $A(5)=100/e\approx36.79$. In ML, products of small probabilities can underflow; summing their logarithms avoids forming that product. Logs of probabilities are nonpositive; negative log loss reverses the sign.

```python
print("log2(32):", math.log2(32))
print("Decay at t=5:", 100 * math.exp(-0.2 * 5))
probabilities = [0.8, 0.6, 0.9]
log_likelihood = sum(math.log(p) for p in probabilities)
print("Product:", math.prod(probabilities))
print("Log-likelihood:", log_likelihood)
assert math.isclose(math.exp(log_likelihood), math.prod(probabilities))
print("Accurate log near 1:", math.log1p(1e-12))
```

## 9. Scientific notation

Write a nonzero number as $a\times10^k$ with $1\le|a|<10$ and integer $k$.

- $0.000072=7.2\times10^{-5}$.
- $4,500,000=4.5\times10^6$.
- $(3\times10^4)(2\times10^{-7})=6\times10^{-3}$.

Python writes $7.2\times10^{-5}$ as `7.2e-5`. Scientific notation helps compare scales, but does not remove floating point rounding, overflow, or underflow. For example, a sufficiently tiny positive product can become `0.0`.

```python
print(f"Small value: {0.000072:.2e}")
print("Scale product:", 3e4 * 2e-7)
print("Underflow example:", 1e-200 * 1e-200)
print("Log of that mathematical product:", 2 * math.log(1e-200))
```

## 10. Putting it together: prediction errors

For true values $y_i$ and predictions $\hat y_i$, the mean squared error averages squared differences. The root mean squared error has the same units as the target. A relative improvement compares the loss reduction against the old loss, not against the new loss.

```python
actual = [10, 20, 30]
predicted = [12, 18, 33]
squared_errors = [(y - pred)**2 for y, pred in zip(actual, predicted)]
loss = sum(squared_errors) / len(actual)
print("Squared errors:", squared_errors)
print("MSE:", loss, "RMSE:", math.sqrt(loss))
old_loss = 10
print("Loss reduction (%):", 100 * (old_loss - loss) / old_loss)
```
