## 1. Variables, constants, and algebraic expressions

A **variable** represents a quantity that can vary; a **constant** is fixed in the current problem. In $\hat y=wx+b$, $x$ is an input, $\hat y$ a prediction, and $w,b$ are model parameters. During prediction, fitted parameters are held fixed; during training, they vary.

An **expression** such as $3x+2$ has a value after substitution. An **equation**, such as $3x+2=11$, asserts equality. Terms are added or subtracted; coefficients multiply variable parts. In $4x^2-3x+7$, the coefficients are $4,-3$, and the constant term is $7$.

Combine **like terms**: $3x+2x-4=5x-4$; $x+x^2$ cannot be combined this way. The distributive law gives $a(b+c)=ab+ac$; factoring reverses it. Useful identities:

$$(a+b)^2=a^2+2ab+b^2,\qquad a^2-b^2=(a-b)(a+b).$$

**Worked example:** $2(x+3)-x=2x+6-x=x+6$. At $x=4$, this equals 10. Beware: $(a+b)^2$ generally differs from $a^2+b^2$.

```python
x = 4
print("Original:", 2 * (x + 3) - x, "simplified:", x + 6)
w, b = 2.5, -1
print("Prediction:", w * x + b)
```

## 2. Linear equations

A one-variable linear equation has the form $ax+b=c$ with $a\ne0$, giving $x=(c-b)/a$. Perform the same valid operation on both sides to preserve solutions.

**Worked example:** $3x+5=20\Rightarrow3x=15\Rightarrow x=5$. Substitution checks $3(5)+5=20$.

For $ax+b=0$, if $a=0,b=0$, every real $x$ is a solution; if $a=0,b\ne0$, no solution exists. Do not divide by a coefficient until checking whether it is zero.

Two linear equations can be solved by elimination or substitution:
$$2x+y=7,\qquad x-y=2.$$
Adding yields $3x=9$, hence $x=3,y=1$. Geometrically, these lines intersect once. Parallel distinct lines give no solution; identical lines give infinitely many.

**ML connection:** a feature's coefficient describes a change in prediction per unit change in that feature, with other features fixed. Solving an equation may invert a simple model: if $\hat y=2x+3$, the input producing 11 is $x=4$.

```python
a, b, c = 3, 5, 20
solution = (c - b) / a
print("Linear solution:", solution)
x, y = 3, 1
assert 2*x + y == 7 and x - y == 2
print("System solution:", (x, y))
```

## 3. Quadratic equations

A quadratic is $ax^2+bx+c=0$ with $a\ne0$. You can factor, complete the square, or use
$$x=\frac{-b\pm\sqrt{b^2-4ac}}{2a}.$$

The **discriminant** $D=b^2-4ac$ determines the real roots: two distinct roots for $D>0$, one repeated root for $D=0$, and no real roots for $D<0$ (two complex roots).

**Worked example:** $x^2-5x+6=(x-2)(x-3)=0$, so $x=2$ or 3. Completing the square gives $x^2-4x+7=(x-2)^2+3$, so this expression has minimum 3 at $x=2$ and no real zeros.

The graph is a parabola. Its vertex occurs at $x=-b/(2a)$, a minimum if $a>0$ and a maximum if $a<0$. Squared-error losses often form quadratics in simple model parameters.

```python
import math

def real_quadratic_roots(a, b, c):
    # Teaching implementation for modest, well-scaled coefficients.
    # The direct formula can lose precision for extreme coefficients.
    if a == 0:
        raise ValueError("A quadratic needs a nonzero x^2 coefficient")
    discriminant = b*b - 4*a*c
    if discriminant < 0:
        return ()
    if discriminant == 0:
        return (-b / (2*a),)
    root_d = math.sqrt(discriminant)
    return ((-b - root_d) / (2*a), (-b + root_d) / (2*a))

for coefficients in [(1, -5, 6), (1, -4, 4), (1, 0, 1)]:
    print(coefficients, "->", real_quadratic_roots(*coefficients))
assert real_quadratic_roots(1, -5, 6) == (2.0, 3.0)
```

## 4. Inequalities

Inequalities use $<,\le,>,\ge$. Adding or subtracting the same value preserves the direction. Multiplying or dividing by a **negative** value reverses it.

**Worked example:** $-2x+3\le7\Rightarrow-2x\le4\Rightarrow x\ge-2$.

Compound inequalities describe intersections: $1<2x+3\le9$ becomes $-1<x\le3$. An absolute-value constraint $|x-a|\le r$ for $r\ge0$ means $a-r\le x\le a+r$.

For a polynomial inequality, factor and check signs between roots. Example:
$$(x-2)(x-3)\le0\iff x\in[2,3].$$

When clearing a denominator involving a variable, first find where the denominator is zero and determine its sign on each interval. For example, $1/x>0$ requires $x>0$; multiplying by $x$ without knowing its sign loses information.

**ML connection:** probabilities satisfy $0\le p\le1$, and constrained optimization may require parameters $w\ge0$.

```python
candidates = [-3, -2, 0, 4]
for value in candidates:
    print(value, "satisfies -2x+3 <= 7:", -2*value + 3 <= 7)
```

## 5. Functions and their properties

A function assigns **exactly one output** to each allowed input. Write $f:A\to B$, where $A$ is the domain and $B$ the codomain. The **range** is the set of outputs actually attained, a subset of $B$.

| Function | Real domain | Range |
|---|---|---|
| $f(x)=x^2$ | $\mathbb R$ | $[0,\infty)$ |
| $g(x)=1/x$ | $\mathbb R\setminus\{0\}$ | $\mathbb R\setminus\{0\}$ |
| $h(x)=\sqrt x$ | $[0,\infty)$ | $[0,\infty)$ |
| $\ell(x)=\ln x$ | $(0,\infty)$ | $\mathbb R$ |

- A **zero/root** satisfies $f(x)=0$; the vertical intercept is $f(0)$ if defined.
- An increasing function preserves order; a decreasing one reverses it. Specify strict or non-strict when necessary.
- An **even** function satisfies $f(-x)=f(x)$ (example $x^2$); an **odd** function satisfies $f(-x)=-f(x)$ (example $x^3$).
- Composition: $(f\circ g)(x)=f(g(x))$, only where $g(x)$ is in the domain of $f$.
- A function is **injective** if distinct inputs have distinct outputs; **surjective** if its range equals its codomain; **bijective** if both. An inverse on the stated codomain exists for a bijection.

**Worked example:** if $f(x)=2x+3$ and $g(x)=x^2$, then $f(g(2))=11$ while $g(f(2))=49$. The inverse of $f:\mathbb R\to\mathbb R$ is $f^{-1}(y)=(y-3)/2$.

The function $x^2$ on all real numbers has no inverse function because $2$ and $-2$ have the same output. Restricting its domain and codomain to $[0,\infty)$ gives inverse $\sqrt x$. The notation $f^{-1}$ means inverse, not reciprocal $1/f$.

```python
def f(x):
    return 2*x + 3

def g(x):
    return x*x

def inverse_f(y):
    return (y - 3) / 2

print("f(g(2)):", f(g(2)), "g(f(2)):", g(f(2)))
assert inverse_f(f(8)) == 8
```

## 6. Polynomial functions

A polynomial is $p(x)=a_0+a_1x+\cdots+a_dx^d$, where exponents are nonnegative integers and $a_d\ne0$. Its degree is $d$. Nonzero constants have degree zero; the zero polynomial's degree is usually left undefined.

Examples: $3x+1$ (linear), $x^2-2x$ (quadratic), $2x^3-x+4$ (cubic). The functions $1/x$, $\sqrt x$, and $2^x$ are not polynomials in $x$.

A degree-$d$ nonzero polynomial has at most $d$ distinct real roots. Its leading term controls behavior for large $|x|$. All polynomials are continuous and differentiable on $\mathbb R$.

**Worked example:** $p(x)=2x^3-3x^2+x-5$ has degree 3 and $p(2)=16-12+2-5=1$.

Polynomial regression can use $\hat y=b+w_1x+w_2x^2$. It is nonlinear in the input $x$ but linear in the fitted coefficients $b,w_1,w_2$. High powers can magnify input scales and produce unstable extrapolation.

```python
def polynomial(x, coefficients):
    # Coefficients are in ascending order: constant, x, x^2, ...
    return sum(coef * x**power for power, coef in enumerate(coefficients))

print("Cubic at 2:", polynomial(2, [-5, 1, -3, 2]))
print("Quadratic predictions:", [polynomial(x, [1, 2, 0.5]) for x in [-2, -1, 0, 1, 2]])
```

## 7. Exponential and logarithmic functions

An exponential $f(x)=a^x$ has $a>0,a\ne1$. Its domain is $\mathbb R$, its range $(0,\infty)$, and $f(0)=1$. It grows for $a>1$ and decays for $0<a<1$. A polynomial varies the base; an exponential varies the exponent: $x^2$ and $2^x$ are different functions.

The inverse is $\log_a x$, defined for $x>0$ with range $\mathbb R$. The logarithm passes through $(1,0)$. The equation $a^x=c$ with $c>0$ gives $x=\ln c/\ln a$.

**Worked examples:** $3\times2^t=24\Rightarrow2^t=8\Rightarrow t=3$.
$\ln(x-1)=2\Rightarrow x=1+e^2$, with the required domain $x>1$.

For continuous growth $A_0e^{kt}$ with $k>0$, doubling time solves $e^{kt}=2$ and equals $\ln2/k$.

**ML connection:** the sigmoid $\sigma(z)=1/(1+e^{-z})$ maps real scores into $(0,1)$. Its inverse is the logit $\ln(p/(1-p))$, defined only for $0<p<1$. Endpoint probabilities correspond to limiting infinite scores.

```python
def sigmoid(z):
    # Equivalent branches avoid exp overflow for large-magnitude scores.
    if z >= 0:
        return 1 / (1 + math.exp(-z))
    exp_z = math.exp(z)
    return exp_z / (1 + exp_z)

for z in [-1000, -2, 0, 2, 1000]:
    print(z, "->", sigmoid(z))
p = 0.8
logit = math.log(p / (1-p))
print("Logit(0.8):", logit)
assert math.isclose(sigmoid(logit), p)
# Finite precision can round a sigmoid output to 0 or 1 at extreme scores.
```

## 8. Modeling example: a loss is a function of parameters

Suppose the model is $\hat y=wx$ (no intercept) and the data are $(1,2),(2,4),(3,6)$. Its mean squared error is
$$L(w)=\frac{(2-w)^2+(4-2w)^2+(6-3w)^2}{3}
=\frac{14}{3}(w-2)^2.$$
The squared factor is nonnegative, so the minimum is zero at $w=2$. This algebraic argument finds the best parameter without calculus. Later we will find it using derivatives.

```python
xs, ys = [1, 2, 3], [2, 4, 6]
def loss(w):
    return sum((y - w*x)**2 for x, y in zip(xs, ys)) / len(xs)

for w in [0, 1, 2, 3, 4]:
    print(f"w={w}, MSE={loss(w):.6f}")
assert loss(2) == 0
```
